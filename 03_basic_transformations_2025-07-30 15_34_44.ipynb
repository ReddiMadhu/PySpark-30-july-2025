13. How to find the numbers that are multiples of 3 from a column?

```python
from pyspark.sql.functions import rand
# Generate a DataFrame with a single column "id" with 10 rows
df = spark.range(10)
# Generate a random float between 0 and 1, scale and shift it to get a random integer between 1 and 10
df.show()
df = df.withColumn("random", ((rand(seed=42) * 10)+1).cast("int"))
# Show the DataFrame
df.show()
```

```python
df.select("id","random").filter(df["random"]%3==0).show()
```

```python
from pyspark.sql.functions import when

df = df.withColumn("is_mutliple_of_3", when(df["random"] % 3 == 0, 1).otherwise(0))
display(df)
```

How to extract items at given positions from a column?

```python
from pyspark.sql.functions import rand
# Generate a DataFrame with a single column "id" with 10 rows
df = spark.range(10)
# Generate a random float between 0 and 1, scale and shift it to get a random integer between 1 and 10
df = df.withColumn("random", ((rand(seed=42) * 10) + 1).cast("int"))
# Show the DataFrame
df.show()
pos = [0, 4, 8, 5]
```

```python
df = df.filter(df["id"].isin(pos))
display(df)
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number, monotonically_increasing_id
```

```python
w=Window.orderBy(monotonically_increasing_id())
df=df.withColumn("index",row_number().over(w)-1)
df.show()
df=df.filter(df["index"].isin(pos))
df.show()
```
